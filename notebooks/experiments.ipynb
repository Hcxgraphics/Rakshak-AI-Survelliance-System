```python
from PIL import Image
import os
from glob import glob

def resize_folder_images(folder, size=(256, 256)):
    img_paths = glob(os.path.join(folder, '*.jpg')) + glob(os.path.join(folder, '*.png'))
    for img_path in img_paths:
        try:
            img = Image.open(img_path)
            img = img.resize(size, Image.BILINEAR)
            img.save(img_path)
        except Exception as e:
            print(f"Error resizing {img_path}: {e}")

# Resize all images in train/val/test
base_dir = 'danger-or-police-1'

for split in ['train', 'valid', 'test']:
    resize_folder_images(os.path.join(base_dir, split, 'images'))
```

```python
import os
import shutil
from glob import glob

# Adjust paths to your dataset structure
train_img_dir = "danger-or-police-1/train/images"
train_lbl_dir = "danger-or-police-1/train/labels"

# Number of duplicates to create for each police image
DUPLICATE_FACTOR = 3  # Creates 3x more copies

# Identify label files that contain 'police' (class index based on your YAML)
police_class_index = None
names = model.names  # Assuming model is already loaded
for idx, cls in names.items():
    if cls.lower() == 'police':
        police_class_index = str(idx)
        break

assert police_class_index is not None, "Police class not found in model.names"

# Find all label files that include the police class index
police_labels = []
for label_path in glob(os.path.join(train_lbl_dir, '*.txt')):
    with open(label_path, 'r') as f:
        lines = f.readlines()
        if any(line.startswith(police_class_index + ' ') for line in lines):
            police_labels.append(label_path)

print(f"Found {len(police_labels)} police label files")

# Duplicate each police label + image
for label_path in police_labels:
    image_path = label_path.replace('/labels/', '/images/').replace('.txt', '.jpg')
    if not os.path.exists(image_path):
        image_path = label_path.replace('/labels/', '/images/').replace('.txt', '.png')
        if not os.path.exists(image_path):
            continue  # Skip if image not found

    for i in range(DUPLICATE_FACTOR):
        # Create new filenames
        base = os.path.basename(label_path).replace('.txt', '')
        new_lbl = os.path.join(train_lbl_dir, f"{base}_dup{i}.txt")
        new_img = os.path.join(train_img_dir, f"{base}_dup{i}.jpg")
        
        # Copy files
        shutil.copyfile(label_path, new_lbl)
        shutil.copyfile(image_path, new_img)

print(f"✅ Oversampling complete: {DUPLICATE_FACTOR}x more 'police' examples added.")

```
